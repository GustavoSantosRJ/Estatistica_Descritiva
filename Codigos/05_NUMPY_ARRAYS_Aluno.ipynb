## Introdução ao NumPy
O NumPy (abreviação de Numerical Python) é a biblioteca mais fundamental para cálculos e ciência de dados em Python.  

Para entender o NumPy, imagine que uma lista comum do Python é uma caixa onde você pode guardar qualquer coisa bagunçada: meias, sapatos, brinquedos. Já o NumPy funciona como uma bandeja de ovos milimetricamente organizada: cada espaço tem exatamente o mesmo formato, guarda o mesmo tipo de dado e fica perfeitamente alinhado na memória do computador.

### História e Aplicações
Criado em 2005 por Travis Oliphant para unificar projetos anteriores (Numeric e Numarray), o NumPy tornou-se a base de quase todo o ecossistema de dados em Python (como Pandas, SciPy, Scikit-learn e TensorFlow). Ele é amplamente usado em:

- Inteligência Artificial e Ciência de Dados.

- Engenharia e simulações físicas.

- Processamento de imagens (fotos digitais são grades numéricas de pixels).

- Finanças e modelagem de risco.

### Computação Numérica e Processamento Vetorizado
No Python tradicional, se você quiser somar 2 a todos os números de uma lista, precisa fazer um laço de repetição (for), visitando item por item. Isso é lento.

O NumPy introduz o processamento vetorizado: ele aplica uma operação a centenas de milhares de números ao mesmo tempo, utilizando rotinas escritas em C/Fortran nos bastidores.

## Analogia:  
Imagine uma transportadora.  
Uma lista Python entrega caixas uma por uma.  
O NumPy entrega centenas de caixas em um caminhão.  
Por isso ele é mais rápido.

Para maiores informações consulte: [https://numpy.org/devdocs/ ](https://numpy.org/doc/stable/index.html)

## Array Bidimensional (2D)

### Array Multidimensional (3D ou mais)  
Pense em um array 3D como um cubo ou uma pilha de folhas de papel. Cada folha é uma matriz 2D.  
É muito comum no processamento de imagens coloridas (onde temos: altura, largura e 3 canais de cor: Vermelho, Verde e Azul).

```python
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("Dimensões:", arr_3d.ndim)  # 3
print("Formato (shape):", arr_3d.shape)  # (2, 2, 2)
```

## Comparando os Formatos

```python
import numpy as np

# Forma tradicional em Python (com loop for)
lista_python = [1, 2, 3, 4]
resultado_python = [x + 2 for x in lista_python] # forma com list comprehension, mas ainda é um loop por baixo dos panos

# No NumPy (vetorizado, direto e muito mais rápido)
array_numpy = np.array([1, 2, 3, 4])
resultado_numpy = array_numpy + 2  # Soma 2 em todos de uma vez

print("Python tradicional:", resultado_python)
print("NumPy vetorizado:", resultado_numpy)
```

### Atividade 2
Dada a lista de preços precos = [100, 250, 400], converta-a em um array NumPy e aplique um desconto de 10% em todos os valores de uma só vez (sem usar laços for).  

```python
import numpy as np

precos_python = [100, 250, 400]
precos_numpy = np.array(precos_python)
precos_numpy = precos_numpy * 0.9

print("Preços originais (lista do Python):  ", precos_python)
print("Preços com Desconto (array do Numpy):", precos_numpy)
```

```python
saida_formatada = np.array2string(precos_numpy, formatter={'float_kind': lambda valor: f'{valor:,.2f}'})
print(saida_formatada)
print('Posições dos preços:', np.arange(len(precos_numpy)))
```

## Operações Matemáticas
O NumPy oferece funções estatísticas e matemáticas rápidas e prontas para uso.

- Soma (sum): totaliza os valores.

- Média (mean): valor central do conjunto.

- Máximo (max): maior valor encontrado.

- Mínimo (min): menor valor encontrado.

- Desvio Padrão (std): mede o quanto os valores variam em torno da média (dispersão).

```python
valores = np.array([10, 20, 30, 40, 50])

print("Soma:", np.sum(valores))          # 150
print("Média:", np.mean(valores))        # 30.0
print("Máximo:", np.max(valores))        # 50
print("Mínimo:", np.min(valores))        # 10
print("Desvio padrão:", np.std(valores)) # ~14.14 
# representa a dispersão dos valores em relação à média
```

*Dica:* Essas funções também funcionam como métodos do próprio array: valores.mean(), valores.max(), etc.

Filtros no NumPy
Para selecionar elementos específicos em um array sem usar laços ou condições complexas, o NumPy utiliza máscaras booleanas.

Operadores Relacionais
Ao comparar um array com um valor, o NumPy compara cada elemento individualmente e retorna um array de valores True e False.

- `> (maior que)`

- `< (menor que)`

- `>= (maior ou igual)`

- `<= (menor ou igual)`

- `== (igual a)`

- `!= (diferente de)`

```python
idades = np.array([12, 17, 18, 25, 40])

# Criação da máscara
mascara_maioridade = idades >= 18
print(mascara_maioridade)
# Saída: [False False  True  True  True]
```

### Máscaras Booleanas (Filtragem)  
Quando passamos essa máscara entre colchetes (array[mascara]), o NumPy filtra e preserva apenas os elementos onde o valor é True.

```python
idades = np.array([12, 17, 18, 25, 40])

# Filtrando os dados
maiores_de_idade = idades[idades >= 18]
print('Maiores de idade:', maiores_de_idade)  # [18, 25, 40]

# Múltiplas condições: use & (E) ou | (OU) com parênteses obrigatórios
jovens_adultos = idades[(idades >= 18) & (idades <= 30)]
print('"Xóvens":', jovens_adultos)    # [18, 25]
```

### Atividade 5
Cenário
A DataOrganized recebeu uma planilha contendo o faturamento de diferentes filiais.
Dados:

faturamento = np.array([12000,15000,18000,21000,25000,30000,35000,40000])

#### Atividades
Parte 1  
Calcular:

- Soma
- Média
- Máximo
- Mínimo

Parte 2  
Aplicar reajuste de: 12% em todas as filiais.

Parte 3  
Exibir apenas faturamentos acima de: R$ 25.000


Parte 4  
Determinar: Quantidade de filiais


```python
ar_faturamento = np.array([12000,15000,18000,21000,25000,30000,35000,40000])

# Imprime algumas estatísticas (parte 1)
print(f'Total Faturado:    {ar_faturamento.sum():10,.2f}')
print(f'Faturamento Médio: {ar_faturamento.mean():10,.2f}')
print(f'Maior Faturamento: {ar_faturamento.max():10,.2f}')
print(f'Menor Faturamento: {ar_faturamento.min():10,.2f}')

ar_ajustado = ar_faturamento * 1.12 # Reajusta faturamento em 12% (parte 2)

# Imprime todos os faturamento ajustados acima de R$25.000 (parte 3)
ar_ajustado = ar_ajustado[ar_ajustado > 25000]
print('\nFaturamentos acima de R$25 mil (com valores ajustados):')
print(np.array2string(ar_ajustado, formatter={'float_kind': lambda valor: f'{valor:10,.2f}'}))

# Imprime quantidade de filiais (parte 4)
print('\nQuantidade de Filiais:', len(ar_faturamento))
```
