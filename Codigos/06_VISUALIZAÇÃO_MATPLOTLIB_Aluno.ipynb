## 📊 Tipos Comuns de Gráficos (Exemplos Práticos)

Para construir a maioria dos gráficos, utiliza-se a interface de alto nível pyplot (geralmente importada sob o alias plt). Veja como criar as visualizações mais tradicionais:

1. Gráfico de Linhas (plt.plot)

    Ideal para demonstrar tendências ao longo do tempo ou sequências contínuas.

```python
import matplotlib.pyplot as plt

# Dados de exemplo
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai'] # Meses do ano
vendas = [10, 15, 7, 19, 23] # valores de vendas correspondentes a cada mês

# Criar o gráfico
plt.figure(figsize=(8, 4)) # Define o tamanho da imagem em polegadas
plt.plot(meses, vendas, marker='o', color='purple', linestyle='--') # Adiciona marcadores, cor e estilo de linha

# Customização
plt.title('Evolução de Vendas Mensais') # Adiciona título ao gráfico
plt.xlabel('Meses') # Adiciona rótulo ao eixo x
plt.ylabel('Vendas (em unidades)') # Adiciona rótulo ao eixo y
plt.grid(True) # Adiciona linhas de grade

# Exibir resultado
plt.show()
```

2. Gráfico de Barras (plt.bar ou plt.barh)

    Excelente para comparar quantidades entre diferentes categorias discretas.

```python
categorias = ['Produto A', 'Produto B', 'Produto C']
quantidades = [45, 28, 60]

plt.figure(figsize=(4, 4)) # Define o tamanho da imagem em polegadas
plt.bar(categorias, quantidades, color=['#4af', '#f4a', '#af4']) # Cria o gráfico de barras com cores personalizadas podemos usar o nome em inglês ou hexadecimal
plt.title('Estoque por Categoria de Produto')
plt.ylabel('Quantidade')
plt.show()
```

3. Gráfico de Dispersão (plt.scatter)

    Utilizado para analisar a correlação ou a distribuição espacial entre duas variáveis numéricas

```python
import numpy as np

# Gerar dados aleatórios correlacionados
x = np.random.rand(100)
y = x * 2 + np.random.normal(0, 0.2, 100)

plt.scatter(x, y, alpha=0.6, color='teal')
plt.title('Análise de Correlação entre X e Y')
plt.xlabel('Variável X')
plt.ylabel('Variável Y')
plt.show()
print(f'Desvio Padrão: {x.std():,.4f}')
```

4. Histograma (plt.hist)

    Mostra a distribuição de frequência de um conjunto de dados.

```python
dados_alturas = np.random.normal(170, 10, 250)

plt.hist(dados_alturas, bins=20, color='gray', edgecolor='black') # Cria o histograma - bins define o número de barras, edgecolor define a cor da borda das barras
plt.title('Distribuição de Alturas da População')
plt.xlabel('Altura (cm)')
plt.ylabel('Frequência')
plt.show()
```

5.O Boxplot (ou Diagrama de Caixa)  


    É uma ferramenta visual essencial na estatística e ciência de dados. Ele serve para mostrar a distribuição, a dispersão e a simetria de um conjunto de dados, além de identificar de forma instantânea os valores discrepantes (outliers).

## 💻 Exemplo Prático em Python (Matplotlib + Seaborn)

Para este exemplo, simulamos as notas de uma prova realizada por três turmas diferentes (Turma A, Turma B e Turma C) para avaliar o desempenho de cada uma.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Configurando o estilo visual para ficar mais moderno
sns.set_theme(style="whitegrid")

# Criando dados fictícios de notas (de 0 a 100)
np.random.seed(42)
notas_A = np.random.normal(loc=70, scale=10, size=100)
notas_B = np.random.normal(loc=55, scale=15, size=100)
notas_C = np.random.normal(loc=78, scale=8, size=100)

# Adicionando alguns outliers propositais na Turma C (notas muito baixas)
notas_C = np.append(notas_C, [20, 25, 30])

dados = [notas_A, notas_B, notas_C]
labels = ['Turma A', 'Turma B', 'Turma C']

# Criando a figura do Matplotlib
plt.figure(figsize=(10, 6))

# Construindo o Boxplot
# patch_artist=True permite colorir o interior das caixas
box = plt.boxplot(
    dados,
    tick_labels=labels,
    patch_artist=True,
    showmeans=True,
    flierprops=dict(
        marker='o',
        markerfacecolor='red',
        markersize=8,
        linestyle='none'
    )
)

# Customizando as cores das caixas
cores = ['#4C72B0', '#DD8452', '#55A868']
for patch, cor in zip(box['boxes'], cores):
    patch.set_facecolor(cor)
    patch.set_alpha(0.7)

# Alterando a cor da linha da mediana para destacar
for median in box['medians']:
    median.set(color='black', linewidth=2)

# Adicionando títulos e rótulos
plt.title('Distribuição de Notas por Turma', fontsize=16, pad=15)
plt.xlabel('Turmas', fontsize=12)
plt.ylabel('Notas da Prova (0 a 100)', fontsize=12)

# Exibindo o gráfico
plt.show()
```


📊 Como Ler e Interpretar um Boxplot

O Boxplot resume cinco números estatísticos principais (conhecidos como o Resumo dos Cinco Números) e aponta os valores extremos.

*A imagem  a seguir mostra a anatomia de um boxplot com:*

- Q1 = 40 
- Mediana (Q2) = 55
- Q3 = 70
- IQR = 30
- Limite inferior (Q1 - 1.5 × IQR) = -5
- Limite superior (Q3 + 1.5 × IQR) = 115
- Outliers inferiores: -35 e -20
- Outliers superiores: 130 e 142

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

# Estatísticas do boxplot
stats = [{
    'med': 55,
    'q1': 40,
    'q3': 70,
    'whislo': -5,
    'whishi': 115,
    'fliers': [-35, -20, 130, 142]
}]

# Dados da tabela
table_data = [
    ["Q1", "40"],
    ["Mediana (Q2)", "55"],
    ["Q3", "70"],
    ["IQR", "30"],
    ["Limite inferior", "-5"],
    ["Limite superior", "115"],
    ["Outliers inferiores", "-35, -20"],
    ["Outliers superiores", "130, 142"]
]

# Figura com duas áreas
fig = plt.figure(figsize=(14, 8))
gs = GridSpec(1, 2, width_ratios=[1.2, 2.5]) # Define a proporção de largura entre a tabela e o boxplot

# ==========================
# TABELA (ESQUERDA)
# ==========================
ax_table = fig.add_subplot(gs[0, 0]) # Adiciona um subplot para a tabela
ax_table.axis('off') # Desativa os eixos para a área da tabela

table = ax_table.table(
    cellText=table_data,
    colLabels=["Métrica", "Valor"],
    loc='center',
    cellLoc='left',
    colColours=['#2A6FDB', '#2A6FDB']
)

table.auto_set_font_size(False) #   
table.set_fontsize(10) 
table.scale(1.2, 2) # salva a proporção da tabela para melhor visualização (1.2)

# Estilização da tabela
for (row, col), cell in table.get_celld().items(): # note que o row e col começam do 0, então a primeira linha é o cabeçalho

    # Cabeçalho
    if row == 0:  
        cell.set_text_props(color='white', weight='bold')
        cell.set_facecolor('#2A6FDB')
    else:
        if row % 2 == 0:
            cell.set_facecolor('#F5F5F5')
        else:
            cell.set_facecolor('white')

ax_table.set_title(
    "Resumo Estatístico",
    fontsize=13,
    fontweight='bold',
    pad=20 # pad é 
)

# ==========================
# BOXPLOT (DIREITA)
# ==========================
ax = fig.add_subplot(gs[0, 1])

ax.bxp(
    stats,
    showfliers=True,
    patch_artist=True,
    boxprops=dict(
        facecolor="#C7D9F5",
        edgecolor="#2A6FDB",
        linewidth=2
    ),
    medianprops=dict(
        color="#D7301F",
        linewidth=2.8
    ),
    whiskerprops=dict(
        color="dimgray",
        linestyle="--",
        linewidth=1.8
    ),
    capprops=dict(
        color="dimgray",
        linewidth=2
    ),
    flierprops=dict(
        marker='o',
        markerfacecolor='#D7301F',
        markeredgecolor='black',
        markersize=7
    )
)

ax.set_title(
    "Anatomia de um Boxplot (Gráfico de Caixa)",
    fontsize=14,
    pad=15
)

ax.set_ylabel("Valores Numéricos Crescentes")
ax.set_xticks([])
ax.set_ylim(-50, 160) # Define os limites do eixo y

# Anotações
ax.text(1.12, 70,
        "Q3 (Terceiro Quartil / Percentil 75)",
        va='center')

ax.text(1.13, 55,
        "Mediana (Q2 / Percentil 50)",
        color="#D7301F",
        va='center')

ax.text(1.12, 40,
        "Q1 (Primeiro Quartil / Percentil 25)",
        va='center')

ax.text(1.02, 117,
        "Máximo Local (Q3 + 1.5 × IQR)")

ax.text(1.02, -7,
        "Mínimo Local (Q1 - 1.5 × IQR)")

ax.text(1.02, 135,
        "Outliers (Valores Discrepantes)",
        color="#D7301F")

ax.text(1.02, -28,
        "Outliers (Valores Discrepantes)",
        color="#D7301F")

# IQR
ax.annotate(
    "",
    xy=(0.82, 70),
    xytext=(0.82, 40),
    arrowprops=dict(
        arrowstyle="]-[",
        lw=1.5,
        color="black"
    )
)

ax.text(
    0.72,
    55,
    "IQR (Intervalo Interquartil)\nContém 50% dos dados",
    ha="right",
    va="center"
)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

Entenda o papel de cada elemento anatômico do gráfico:

1. A Caixa Central (O "Corpo" do Gráfico)

    • Linha Inferior da Caixa (Q₁ - 1º Quartil): Representa o percentil 25%. Significa que 25% dos dados estão abaixo desse valor.   

    • Linha do Meio da Caixa (Q₂ - Mediana): Divide os dados exatamente ao meio. 50% dos dados estão abaixo dessa linha e 50% estão acima. Se ela   não estiver bem no centro da caixa, indica que os dados são assimétricos.  

    • Linha Superior da Caixa (Q₃ - 3º Quartil): Representa o percentil 75%.  Significa que 75% dos dados estão abaixo desse valor (ou que 25% estão acima).  

    • O tamanho da caixa (IQR): Chamado de Intervalo Interquartil (IQR = Q₃ - Q₁). Ele engloba exatamente os 50% centrais de todos os seus dados.   Quanto maior a caixa, mais dispersos (variáveis) são os dados centrais.

2. As Linhas Verticais (Os "Whiskers" ou Bigodes)  

    As linhas estendidas para cima e para baixo mostram a variabilidade fora dos 50% centrais. Elas não vão necessariamente até o valor máximo absoluto, mas sim até um limite estatístico seguro:  

    • Limite Superior: Calculado como \(Q_3 + (1,5 \times IQR)\). É o valor máximo esperado antes de uma nota ser considerada anormal. 

    • Limite Inferior: Calculado como \(Q_1 - (1,5 \times IQR)\). É o valor mínimo esperado.  


3. Os Pontos Isolados (Outliers / Valores Discrepantes)  

    • Qualquer dado que fique além dos limites dos bigodes é desenhado como um ponto individual (no código acima, configuramos para aparecer em vermelho).  

    • Eles representam eventos raros, erros de digitação ou casos muito excepcionais na amostra.
