## Bloco 1: O que é Big Data de Verdade? (Os 5 Vs)

Muitas pessoas acham que Big Data significa apenas "um arquivo grande" ou "uma grande base de dados". Na prática corporativa, Big Data ocorre quando os dados superam a capacidade de processamento das ferramentas tradicionais em **cinco dimensões (os 5 Vs)**:

| Dimensão (V) | O que significa na prática? | Exemplo do Dia a Dia |
| :--- | :--- | :--- |
| **1. Volume** | Quantidade massiva de dados gerados (Gigabytes, Terabytes, Petabytes). | Milhões de pagamentos mensais do Novo Bolsa Família ou notas fiscais de uma rede nacional. |
| **2. Velocidade** | Rapidez com que os dados nascem e precisam ser processados. | Transações de cartão de crédito ou PIX acontecendo a cada milissegundo. |
| **3. Variedade** | Diferentes formatos chegando ao mesmo tempo. | Planilhas CSV, bancos SQL, textos, imagens, áudios e registros de sensores. |
| **4. Veracidade** | Grau de confiança e limpeza dos dados. | Dados duplicados, campos vazios (`null`), CPFs digitados errados que precisam de auditoria. |
| **5. Valor** | Capacidade de transformar essa montanha de dados em decisão útil. | Descobrir fraudes, prever estoque ou otimizar investimentos públicos. |

## Bloco 2: Arquitetura Corporativa — Engenheiro vs. Cientista de Dados

Dentro de uma empresa de tecnologia, ninguém faz tudo sozinho. Existe uma **arquitetura dividida em camadas**, onde cada profissional tem ferramentas e limites de máquina bem definidos (como destaca o nosso Plano de Curso):

### 1. A Ótica Corporativa (Onde atua o Engenheiro de Dados)
- **Ambiente:** Nuvem (AWS, Azure, Google Cloud) e *clusters* (dezenas ou centenas de servidores conectados em rede).
  
- **Tecnologia:** **Computação Distribuída e Paralelismo Real** (ex.: Apache Spark, Hadoop, Databricks). Se um arquivo tem 500 GB, o Engenheiro divide esse arquivo em 50 pedaços e manda 50 computadores processarem ao mesmo tempo.
  
- **Missão:** Construir os dutos de dados, garantir segurança, governança e entregar bases organizadas.

### 2. A Ótica da Estação de Trabalho (Onde atua o Cientista / Analista de Dados)
- **Ambiente:** A sua própria máquina local (notebook ou workstation com 8 GB, 16 GB ou 32 GB de RAM e 4 a 16 núcleos de processador).
  
- **Tecnologia:** **Pseudo-paralelismo e Multithreading** (ex.: **Polars**, DuckDB). Como você não tem 50 servidores na sua mesa, precisa extrair 100% da potência dos núcleos do seu próprio processador e economizar cada megabyte de memória RAM.
  
- **Missão:** Explorar, testar hipóteses estatísticas, criar modelos preditivos e gerar inteligência de negócio.

### A Esteira de Produção (Pipeline de Dados)
Todo projeto profissional segue 4 etapas separadas (separar o pré-processamento do processamento evita travar a máquina na hora da análise):

`[1. Pré-processamento (Limpeza/Compactação)]` $\rightarrow$ `[2. Processamento (Cálculos/Estatística)]` $\rightarrow$ `[3. Carga (Banco/Parquet)]` $\rightarrow$ `[4. Apresentação (Power BI/Relatórios)]`

## Bloco 3: Por que o Polars é Tão Diferente do Pandas? (Muito Além da Velocidade!)

A diferença entre Pandas e Polars não é apenas "correr mais rápido" — é a **arquitetura interna** de como os dados são guardados na memória RAM:

### Analogia do Armário de Arquivos vs. Caixas Organizadoras
1. **Como o Pandas guarda dados (Orientado a Linhas / NumPy legado):**
   - Guarda cada linha inteira junta e cria um **índice pesado** para cada linha. Para ler apenas uma coluna de preços, muitas vezes precisa carregar informações vizinhas e criar cópias na memória RAM (dobrando ou triplicando o consumo de memória).
   - Usa apenas **1 núcleo do processador** por vez (*Single-thread*), deixando os outros núcleos ociosos.
2. **Como o Polars guarda dados (Apache Arrow — Orientado a Colunas):**
   - **Memória Colunar (Apache Arrow):** Cada coluna fica guardada em um bloco contínuo e compactado na RAM. O Polars chega a consumir **3x a 5x menos memória RAM** que o Pandas para o mesmo arquivo!
   - **Sem Índice Desnecessário:** Não gasta memória criando coluna de índice (`0, 1, 2...`).
   - **Motor em Rust + Multithreading:** Divide automaticamente as operações entre **todos os núcleos da CPU** ao mesmo tempo.
   - **Expressões Paralelas:** Permite calcular 5 ou 10 métricas diferentes simultaneamente dentro de um único `.agg()` ou `.with_columns()`, enquanto o Pandas calcula uma coluna de cada vez.
   - **Tipagem Estrita (Segurança):** No Pandas, um único texto perdido no meio de uma coluna numérica transforma a coluna inteira em `object` (texto), quebrando contas matemáticas lá na frente. O Polars valida os tipos desde o início.
   - Link: https://docs.pola.rs/api/python/stable/reference/index.html

#### FLUXO DE TRABALHO: 



````python
# Não Rodar é só um exemplo de como ler um arquivo CSV e realizar operações de agregação com Polars

import polars as pl

df = pl.read_csv("sales.csv") # leitura do arquivo CSV 

result = (    
    df.with_columns(  # adicionando uma nova coluna "total" que é o produto de "price" e "quantity"
        (pl.col("price") * pl.col("quantity")).alias("total")  
    )  
    .groupby("store")  # agrupando os dados por "store"
    .agg(pl.col("total").sum())  #  calculando a soma da coluna "total" para cada grupo
    .sort("total", descending=True)  # ordenando os resultados pela coluna "total" em ordem decrescente
)  

print(result)
````

## Exemplo 1 (Nível Básico): Estrutura, Tipagem e Memória no Polars

Vamos começar criando uma tabela pequena para observar como o Polars organiza os dados, exibe os tipos de cada coluna e calcula o tamanho exato ocupado na memória RAM (`.estimated_size()`).

````python
# Passo 1: Importa a biblioteca Pandas com o apelido padrão 'pd'
import pandas as pd

# Passo 2: Cria um DataFrame com dados de 6 pedidos de uma loja
df_pedidos = pd.DataFrame({
    # Coluna com os identificadores únicos dos pedidos
    'ID_Pedido': [101, 102, 103, 104, 105, 106],

    # Coluna com os nomes dos clientes
    'Cliente': ['Marina', 'Carlos', 'Fernanda', 'Rafael', 'Juliana', 'Lucas'],

    # Coluna que indica a região de cada cliente
    'Regiao': ['Sul', 'Sudeste', 'Sul', 'Nordeste', 'Sudeste', 'Norte'],

    # Coluna com o valor de cada pedido em reais
    'Valor_BRL': [450.00, 1200.50, 89.90, 3400.00, 560.00, 190.00],

    # Coluna booleana que informa se o pedido foi entregue
    'Entregue': [True, True, False, True, True, False]
})

# Exibe o DataFrame completo na tela
print(df_pedidos)

# Calcula o espaço total ocupado pelo DataFrame na memória RAM
# deep=True inclui o consumo real das colunas de texto (strings)
memoria_bytes = df_pedidos.memory_usage(deep=True).sum()

# Converte o valor de bytes para kilobytes (KB)
memoria_kb = memoria_bytes / 1024

# Exibe o espaço ocupado na memória RAM formatado com duas casas decimais
print(f"Espaço ocupado na RAM: {memoria_kb:.2f} KB")
````

````python
# Passo 1: Importa a biblioteca Polars com o apelido oficial 'pl'
import polars as pl

# Passo 2: Cria um DataFrame com dados de 6 pedidos de uma loja
df_pedidos = pl.DataFrame({
    'ID_Pedido': [101, 102, 103, 104, 105, 106],
    'Cliente': ['Marina', 'Carlos', 'Fernanda', 'Rafael', 'Juliana', 'Lucas'],
    'Regiao': ['Sul', 'Sudeste', 'Sul', 'Nordeste', 'Sudeste', 'Norte'],
    'Valor_BRL': [450.00, 1200.50, 89.90, 3400.00, 560.00, 190.00],
    'Entregue': [True, True, False, True, True, False]
})

# Exibe a tabela e o espaço ocupado na memória RAM
print(df_pedidos)
print(f"Espaço ocupado na RAM: {df_pedidos.estimated_size('kb'):.2f} KB")
````

*Leitura Analítica (Exemplo 1):*

Observe o cabeçalho gerado pelo Polars: logo abaixo do nome de cada coluna aparece o tipo exato (`i64` = número inteiro, `str` = texto, `f64` = número decimal, `bool` = verdadeiro/falso) e não existe coluna de índice à esquerda. Além disso, o método `.estimated_size()` permite monitorar exatamente quanto de RAM estamos consumindo — hábito essencial em Big Data.

## Exemplo 2 (Nível Básico-Intermediário): Expressões Paralelas com `select`, `filter` e `with_columns`

### Tabela de Tradução Rápida (Pandas $\rightarrow$ Polars):
| O que você quer fazer | Comando no Pandas | Comando no Polars |
| :--- | :--- | :--- |
| **Escolher colunas** | `df[['Cliente', 'Valor_BRL']]` | `df.select(['Cliente', 'Valor_BRL'])` |
| **Filtrar linhas** | `df[df['Valor_BRL'] > 500]` | `df.filter(pl.col('Valor_BRL') > 500)` |
| **Criar várias colunas juntas** | Uma linha de código por coluna | `df.with_columns([expr1, expr2])` (roda em paralelo!) |

Vamos criar **duas colunas novas ao mesmo tempo** (valor do imposto de 10% e valor líquido) e filtrar apenas os pedidos já entregues.

````python
import polars as pl

# Constrói um pipeline encadeado (uma etapa abaixo da outra)
relatorio_entregues = (
    df_pedidos
    # 1. Filtra apenas pedidos entregues (Entregue == True) E com valor acima de R$ 200
    .filter((pl.col('Entregue') == True) & (pl.col('Valor_BRL') > 200))
    
    # 2. Cria DUAS colunas simultaneamente em paralelo na CPU
    .with_columns([
        (pl.col('Valor_BRL') * 0.10).alias('Imposto_10_pct'),
        (pl.col('Valor_BRL') * 0.90).alias('Valor_Liquido')
    ])
    
    # 3. Seleciona apenas as colunas finais para exibição
    .select(['Cliente', 'Regiao', 'Valor_BRL', 'Imposto_10_pct', 'Valor_Liquido'])
)

print(relatorio_entregues)
````

````python
import polars as pl

# Converte o DataFrame do pandas para Polars, caso necessário.
if not isinstance(df_pedidos, pl.DataFrame):
    df_pedidos = pl.DataFrame(df_pedidos.to_dict(orient="list"))

# Mantém apenas pedidos entregues com valor superior a R$ 200.
relatorio_entregues = (
    df_pedidos
    .filter((pl.col("Entregue") == True) & (pl.col("Valor_BRL") > 200))
    # Calcula o imposto e o valor líquido para cada pedido selecionado.
    .with_columns([
        (pl.col("Valor_BRL") * 0.10).alias("Imposto_10_pct"),
        (pl.col("Valor_BRL") * 0.90).alias("Valor_Liquido")
    ])
    # Seleciona as colunas que serão exibidas no relatório.
    .select(["Cliente", "Regiao", "Valor_BRL", "Imposto_10_pct", "Valor_Liquido"])
)

# Exibe o relatório final.
print(relatorio_entregues)
````

*Leitura Analítica (Exemplo 2):*

No Polars, `pl.col('Valor_BRL')` funciona como um ponteiro direto para a coluna na memória Apache Arrow. Quando passamos uma lista dentro de `.with_columns([...])`, o Polars calcula `Imposto_10_pct` e `Valor_Liquido` **ao mesmo tempo** em núcleos diferentes do processador.

## Exemplo 3 (Nível Intermediário): Estatística Descritiva em Grupo (`group_by` + `agg`)

Agora vamos aplicar as métricas estatísticas que aprendemos nas aulas anteriores (Média, Mediana, Desvio Padrão e Máximo) dentro de um agrupamento do Polars.

````python
# Agrupa os pedidos por Região e calcula 4 indicadores estatísticos simultaneamente
estatisticas_regiao = (
    df_pedidos
    .group_by('Regiao')
    .agg([
        pl.col('Valor_BRL').count().alias('Qtd_Pedidos'),       # Contagem de registros
        pl.col('Valor_BRL').mean().alias('Media_BRL'),          # Média aritmética
        pl.col('Valor_BRL').median().alias('Mediana_BRL'),      # Mediana (Q2)
        pl.col('Valor_BRL').std(ddof=0).alias('Desvio_Padrao')  # Desvio padrão populacional
    ])
    .sort('Media_BRL', descending=True)                         # Ordena da maior média para a menor
)

print(estatisticas_regiao)
````

*Leitura Analítica (Exemplo 3):*

Note que o Polars possui nativamente `.mean()`, `.median()`, `.std(ddof=0)`, `.var(ddof=0)`, `.quantile(0.25)` e `.skew()`. Todas essas operações rodam em paralelo dentro do `.agg([...])` sem precisar fazer um `for` manual coluna por coluna.

## Exemplo 4 (Nível Intermediário): Auditoria de Memória RAM e Tempo — Pandas vs. Polars (2 Milhões de Linhas)

Vamos provar na prática os **dois grandes ganhos** da arquitetura Apache Arrow do Polars sobre uma base de **2.000.000 de registros**:
1. **Economia de Memória RAM** (quantos Megabytes cada biblioteca consome).
2. **Velocidade de Execução** (segundos gastos para filtrar e calcular estatísticas).

````python
import numpy as np
import pandas as pd
import time

# Gera 2 milhões de linhas simulando uma base de transações em larga escala
np.random.seed(42)
n = 2_000_000

dados_2m = {
    'UF': np.random.choice(['RJ', 'SP', 'MG', 'BA', 'PR', 'RS', 'PE', 'CE'], size=n),
    'Tipo_Beneficio': np.random.choice(['Basico', 'Variavel', 'Primeira_Infancia'], size=n),
    'Valor_Parcela': np.random.uniform(150.0, 1200.0, size=n)
}

# Carrega os mesmos dados no Pandas e no Polars
df_pd = pd.DataFrame(dados_2m)
df_pl = pl.DataFrame(dados_2m)

print(f"Quantidade de linhas — Polars: {len(df_pl):,}")
print(f"Quantidade de linhas — Pandas: {len(df_pd):,}".replace(",", "."))

# Compara o consumo exato de memória RAM em Megabytes (MB)
ram_pd_mb = df_pd.memory_usage(deep=True).sum() / (1024 ** 2)
ram_pl_mb = df_pl.estimated_size('mb')

print(f"Consumo de RAM — Pandas: {ram_pd_mb:.2f} MB")
print(f"Consumo de RAM — Polars: {ram_pl_mb:.2f} MB")
print(f"Economia de Memória:     Polars ocupou {ram_pd_mb / ram_pl_mb:.1f}x menos espaço na RAM!")
````

````python
# Compara o tempo para filtrar parcelas > R$ 600 e calcular Média, Mediana e Soma por UF
t0 = time.time()
res_pd = (
    df_pd[df_pd['Valor_Parcela'] > 600]
    .groupby('UF')['Valor_Parcela']
    .agg(['count', 'mean', 'median', 'sum'])
)
tempo_pd = time.time() - t0

t1 = time.time()
res_pl = (
    df_pl
    .filter(pl.col('Valor_Parcela') > 600)
    .group_by('UF')
    .agg([
        pl.col('Valor_Parcela').count().alias('count'),
        pl.col('Valor_Parcela').mean().alias('mean'),
        pl.col('Valor_Parcela').median().alias('median'),
        pl.col('Valor_Parcela').sum().alias('sum')
    ])
)
tempo_pl = time.time() - t1

print(f"Tempo de Execução — Pandas: {tempo_pd:.4f} segundos")
print(f"Tempo de Execução — Polars: {tempo_pl:.4f} segundos")
print(f"Ganho de Velocidade:        Polars foi {tempo_pd / tempo_pl:.1f}x mais rápido!")
````

*Leitura Analítica (Exemplo 4 — Memória + Velocidade):*

Aqui fica evidente por que o Polars é indispensável na estação de trabalho do Cientista de Dados: além de processar os cálculos **muito mais rápido** usando todos os núcleos da CPU, ele consome **muito menos memória RAM** graças ao armazenamento colunar Apache Arrow — evitando que o computador trave quando abrirmos bases reais de vários gigabytes.

### ATIVIDADE 1 (Nível Básico) — Manipulação de Folha de Pagamento com Polars
Crie um `pl.DataFrame` com os dados de 6 colaboradores da **Data Organized**:
- `Colaborador`: `['Alice', 'Bruno', 'Camila', 'Daniel', 'Eduarda', 'Felipe']`
- `Cargo`: `['Analista Jr', 'Engenheiro', 'Analista Pl', 'Cientista', 'Analista Jr', 'Cientista']`
- `Salario_Base`: `[3500.0, 8200.0, 5400.0, 9100.0, 3600.0, 8800.0]`

**Faça:**
1. Exiba o DataFrame e verifique o consumo de memória com `.estimated_size('kb')`.
2. Filtre apenas quem ganha acima de `R$ 5.000,00` e crie duas novas colunas simultaneamente em `.with_columns([...])`: `Vale_Alimentacao` (valor fixo `850.0`, usando `pl.lit(850.0)`) e `Remuneracao_Total` (`Salario_Base + 850.0`).

````python
# Escreva aqui o código da Atividade 1 (Nível Básico):

import polars as pl

# Cria o DataFrame do Polars
df_colaboradores = pl.DataFrame({
    'Colaborador': ['Alice', 'Bruno', 'Camila', 'Daniel', 'Eduarda', 'Felipe'],
    'Cargo': ['Analista Jr', 'Engenheiro', 'Analista Pl', 'Cientista', 'Analista Jr', 'Cientista'],
    'Salario_Base': [3500.0, 8200.0, 5400.0, 9100.0, 3600.0, 8800.0]
})

# Exibe a tabela e o espaço ocupado na memória RAM (atividade 1.1)
print(df_colaboradores)
print(f"Espaço ocupado na RAM: {df_colaboradores.estimated_size('kb'):.2f} KB")

# Cria novo dataframe a partir do original com salários acima de 5mil e colunas novas (atividade 1.2)
df_novo = (df_colaboradores
    .filter(pl.col('Salario_Base') > 5_000)
    .with_columns([
        (pl.lit(850.0)).alias('Vale_Alimentacao'),
        (pl.col('Salario_Base') + 850).alias("Remuneracao_Total")
    ]))

# Exibe novo dataframe
print(df_novo)
print(f"Espaço ocupado na RAM: {df_novo.estimated_size('kb'):.2f} KB")
````

### ATIVIDADE 2 (Nível Intermediário) — Auditoria Estatística em 2 Milhões de Registros
Utilizando o DataFrame `df_pl` de **2 milhões de linhas** criado no Exemplo 4:

**Faça:**
1. Filtre apenas os registros do estado `'RJ'` ou `'SP'` (`pl.col('UF').is_in(['RJ', 'SP'])`) onde `Valor_Parcela` seja maior que `500.0`.
2. Agrupe por `UF` e `Tipo_Beneficio` (`.group_by(['UF', 'Tipo_Beneficio'])`) e calcule:
   - Contagem de beneficiários (`.count().alias('Qtd')`)
   - Média da parcela (`.mean().alias('Media')`)
   - Mediana da parcela (`.median().alias('Mediana')`)
   - Desvio padrão populacional (`.std(ddof=0).alias('DP')`)
3. Ordene pelo maior valor de `Qtd` e cronometre o tempo total da operação.

````python
# Escreva aqui o código da Atividade 2 (Nível Intermediário):
inicio = time.time()

# Filtra registros de Rio e São Paulo com valores maior que 500,00 (atividade 2.1)
df = df_pl.filter((pl.col('Valor_Parcela') > 500), (pl.col('UF') == 'RJ') | (pl.col('UF') == 'SP'))

# Agrupa por UF e tipo do benefício, calculando estatísticas (atividade 2.2)
res_pl = (df
    .group_by('UF', 'Tipo_Beneficio')
    .agg([
        pl.col('Valor_Parcela').count().alias('Qtd'),
        pl.col('Valor_Parcela').mean().alias('Media'),
        pl.col('Valor_Parcela').median().alias('Mediana'),
        pl.col('Valor_Parcela').std(ddof=0).alias('DP')
    ]))

# Imprime resultado em orde decrescente de quantidade e o tempo total da operação (atividade 2.3)
print(res_pl.sort('Qtd', descending=True))
tempo = time.time() - inicio
print(f"Tempo de Execução : {tempo:.4f} segundos")
````
