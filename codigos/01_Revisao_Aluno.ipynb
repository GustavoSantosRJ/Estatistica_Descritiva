Exercício 1: Análise de Vendas
Enunciado:
Crie um programa que analise um conjunto de dados de vendas. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as colunas: Produto, Quantidade, Preço e retornar um DataFrame.

Função calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.

Função total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) e a receita total (soma das receitas). Retorne um novo DataFrame contendo o total de vendas e a receita por produto.

Função filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os produtos cuja receita total é superior a um valor limite passado como argumento.

Função analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.

```python
import pandas as pd

'''
Função le_dados_vendas(nome_arquivo)
Esta função deve ler um arquivo CSV contendo informações de vendas com as colunas: Produto, Quantidade, Preço e retornar um DataFrame.
'''
def le_dados_vendas(nome_arquivo=''):
    if len(nome_arquivo) > 0:
        try:
            df_csv = pd.read_csv(nome_arquivo)
            return df_csv
        except Exception as exc:
            return exc 

'''
Função calcula_receita(df)
Esta função deve receber um DataFrame e calcular a receita total de cada produto (multiplicando Quantidade por Preço).
A receita deve ser adicionada como uma nova coluna chamada Receita.
'''
def calcula_receita(df):
    df['Receita'] = df['Quantidade'] * df['Preco']

'''
Função total_vendas_por_produto(df)
Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) e a receita total.
Retorne um novo DataFrame contendo o total de vendas e a receita por produto.
'''
def total_vendas_por_produto(df):
    df_retorno = (
        df.groupby('Produto')
        [['Quantidade', 'Receita']].sum()
    )
    return df_retorno

'''
Função filtra_produtos_altas_vendas(df, limite)
Esta função deve filtrar o DataFrame para retornar apenas os produtos cuja receita total é superior a um valor limite passado como argumento.
'''
def filtra_produtos_altas_vendas(df, limite=0):
    df_retorno = df.query('Receita > @limite').copy()
    return df_retorno.sort_values(['Receita', 'Produto'], ascending=[False, True])

'''
Função analise_vendas(nome_arquivo)
Esta função deve orquestrar as chamadas das funções anteriores, retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.
'''
def analise_vendas(nome_arquivo):
    df_vendas = le_dados_vendas(nome_arquivo)
    calcula_receita(df_vendas)
    df_vendas_prod = total_vendas_por_produto(df_vendas)
    print(filtra_produtos_altas_vendas(df_vendas_prod, 500))


if __name__ == "__main__":
    
    analise_vendas('D:\\Projetos\\Senac 02_Estatistica\\Data\\vendas.csv')
```

Função le_dados_alunos(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de alunos com as colunas: Nome, Nota1, Nota2, Nota3 e retornar um DataFrame.

Função verifica_aprovacao(df): Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.

Função separa_alunos_aprovados_reprovados(df): Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.

Função analise_notas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.

```python
import pandas as pd

'''
Função le_dados_alunos(nome_arquivo):
Esta função deve ler um arquivo CSV contendo informações de alunos com as colunas: Nome, Nota1, Nota2, Nota3 e retornar um DataFrame.
'''
def le_dados_alunos(nome_arquivo=''):
    if len(nome_arquivo) > 0:
        try:
            df_csv = pd.read_csv(nome_arquivo)
            return df_csv
        except Exception as exc:
            return exc 

'''
Função calcula_media(df):
Esta função deve calcular a média das notas de cada aluno e adicionar uma nova coluna chamada Media ao DataFrame.
'''
def calcula_media(df):
    df['Media'] = (df['Nota1'] + df['Nota2'] + df['Nota3']) / 3

'''
Função verifica_aprovacao(df):
Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.
'''
def verifica_aprovacao(df):
    df['Aprovado'] = (df['Media'] >= 7)

'''
Função separa_alunos_aprovados_reprovados(df):
Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.
'''
def separa_alunos_aprovados_reprovados():
    df_aprovados = df_alunos.query('Aprovado == True')[['Nome', 'Media']]
    df_reprovados = df_alunos.query('Aprovado == False')[['Nome', 'Media']]

'''
Função analise_notas(nome_arquivo):
Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.
'''
def analise_notas(nome_arquivo):
    df_alunos = le_dados_alunos(nome_arquivo)
    calcula_media(df_alunos)
    verifica_aprovacao(df_alunos)
    df_aprovados = df_alunos.query('Aprovado == True')[['Nome', 'Media']]
    df_reprovados = df_alunos.query('Aprovado == False')[['Nome', 'Media']]
    print('Alunos aprovados:')
    print(df_aprovados)
    print('\nAlunos reprovados:')
    print(df_reprovados)


if __name__ == "__main__":
    
    analise_notas('D:\\Projetos\\Senac 02_Estatistica\\Data\\notas.csv')
```

Exercício 3: Análise de Dados de Funcionários
Enunciado:
Crie um programa que analise os dados de funcionários de uma empresa. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.

Função calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.

Função filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.

Função analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.

```python
import pandas as pd

'''
Função le_dados_funcionarios(nome_arquivo):
Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.
]'''
def le_dados_funcionarios(nome_arquivo=''):
    if len(nome_arquivo) > 0:
        try:
            df_csv = pd.read_csv(nome_arquivo)
            return df_csv
        except Exception as exc:
            return exc 

'''
Função calcula_salario_medio(df):
Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um,
retornando um novo DataFrame com o nome do departamento e o salário médio.
'''
def calcula_salario_medio(df):
    df_retorno = df.copy()
    df_retorno = df_retorno.groupby('Departamento')['Salário'].mean()
    return df_retorno

'''
Função filtra_departamentos_alto_salario(df, limite):
Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.
'''
def filtra_departamentos_alto_salario(df, limite=0):
    df_retorno = df.query('Salário > @limite')
    return df_retorno.sort_values(['Salário', 'Departamento'], ascending=[False, True])

'''
Função analise_funcionarios(nome_arquivo):
Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.
'''
def analise_vendas(nome_arquivo):
    df_func = le_dados_funcionarios(nome_arquivo)
    calcula_salario_medio(df_func)
    df_func_dpto = filtra_departamentos_alto_salario(df_func)
    print(filtra_departamentos_alto_salario(df_func_dpto, 5000))


if __name__ == "__main__":
    
    analise_vendas('D:\\Projetos\\Senac 02_Estatistica\\Data\\funcionarios.csv')
```
